## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# DATA CLEANING & PREPROCESSING
# MAKE DATASET ANALYSIS READY
# ============================================================

print("=" * 70)
print("DATA CLEANING & PREPROCESSING")
print("=" * 70)

# Define the target column
TARGET_COLUMN = 'Credit_Score'

# ------------------------------------------------------------
# 1. Create a copy of the original dataset
# ------------------------------------------------------------
# Keeping the original dataframe unchanged allows us to
# compare the cleaned data with the original data if required.
df_clean = data.copy()

print("\nOriginal dataset copied successfully.")


# ------------------------------------------------------------
# 2. Remove leading/trailing spaces from column names
# ------------------------------------------------------------
# This prevents errors caused by accidental spaces in
# column names during analysis and modeling.
df_clean.columns = df_clean.columns.str.strip()

print("Column names cleaned successfully.")


# ------------------------------------------------------------
# 3. Remove leading/trailing spaces from text columns
# ------------------------------------------------------------
# Text columns may contain unwanted spaces such as
# " Good " instead of "Good".
object_columns = df_clean.select_dtypes(
    include=["object"]
).columns

for column in object_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Text columns cleaned successfully.")


# ------------------------------------------------------------
# 4. Standardize known categorical variables
# ------------------------------------------------------------
# Convert categorical values to a consistent format.
categorical_columns_to_clean = [
    "Occupation",
    "Credit_Mix",
    "Payment_of_Min_Amount",
    "Payment_Behaviour",
    "Credit_Score"
]

for column in categorical_columns_to_clean:
    if column in df_clean.columns:
        df_clean[column] = (
            df_clean[column]
            .astype("string")
            .str.strip()
        )

print("Categorical variables standardized successfully.")


# ------------------------------------------------------------
# 5. Convert numeric columns to numeric data types
# ------------------------------------------------------------
# Some financial columns may be stored as object/string
# because of formatting issues in the raw dataset.
numeric_columns = [
    "Age",
    "Annual_Income",
    "Monthly_Inhand_Salary",
    "Num_Bank_Accounts",
    "Num_Credit_Card",
    "Interest_Rate",
    "Num_of_Loan",
    "Delay_from_due_date",
    "Num_of_Delayed_Payment",
    "Changed_Credit_Limit",
    "Num_Credit_Inquiries",
    "Outstanding_Debt",
    "Credit_Utilization_Ratio",
    "Total_EMI_per_month",
    "Amount_invested_monthly",
    "Monthly_Balance"
]

for column in numeric_columns:
    if column in df_clean.columns:
        # errors="coerce" converts invalid numeric values to NaN
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        )

print("Numeric variables converted successfully.")


# ------------------------------------------------------------
# 6. Convert Month to numeric where applicable
# ------------------------------------------------------------
# Month represents the monthly observation number.
if "Month" in df_clean.columns:
    df_clean["Month"] = pd.to_numeric(
        df_clean["Month"],
        errors="coerce"
    )

print("Month variable processed successfully.")


# ------------------------------------------------------------
# 7. Check and remove exact duplicate records
# ------------------------------------------------------------
# Only exact duplicate rows are removed.
# Repeated Customer_ID values are NOT considered duplicates
# because customers may have multiple monthly observations.
duplicate_count = df_clean.duplicated().sum()

if duplicate_count > 0:
    df_clean = df_clean.drop_duplicates().reset_index(drop=True)
    print(f"Removed {duplicate_count:,} exact duplicate records.")
else:
    print("No exact duplicate records found.")


# ------------------------------------------------------------
# 8. Check missing values created during conversion
# ------------------------------------------------------------
# Numeric conversion may create NaN values if invalid text
# was present in a numeric column.
missing_after_cleaning = df_clean.isnull().sum().sum()

print(
    f"Missing values after cleaning: "
    f"{missing_after_cleaning:,}"
)


# ------------------------------------------------------------
# 9. Check target variable
# ------------------------------------------------------------
# The target variable must be available for the classification
# analysis.
if TARGET_COLUMN not in df_clean.columns:
    raise KeyError(
        f"Target column '{TARGET_COLUMN}' was not found."
    )

print(f"Target variable '{TARGET_COLUMN}' is available.")


# ------------------------------------------------------------
# 10. Remove rows with missing target values, if any
# ------------------------------------------------------------
# Rows without a target value cannot be used for supervised
# classification.
target_missing = df_clean[TARGET_COLUMN].isnull().sum()

if target_missing > 0:
    df_clean = df_clean.dropna(
        subset=[TARGET_COLUMN]
    ).reset_index(drop=True)

    print(
        f"Removed {target_missing:,} rows "
        f"with missing target values."
    )
else:
    print("No missing target values found.")


# ------------------------------------------------------------
# 11. Final dataset summary
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("FINAL DATASET SUMMARY")
print("=" * 70)

print(f"Rows after cleaning    : {df_clean.shape[0]:,}")
print(f"Columns after cleaning : {df_clean.shape[1]:,}")
print(
    f"Missing values         : "
    f"{df_clean.isnull().sum().sum():,}"
)
print(
    f"Duplicate rows         : "
    f"{df_clean.duplicated().sum():,}"
)

print("\nData types:")
display(df_clean.dtypes.to_frame(name="Data_Type"))

print("\nFirst 5 cleaned records:")
display(df_clean.head())
````
